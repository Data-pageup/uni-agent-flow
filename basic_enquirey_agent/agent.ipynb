```python
import os
import certifi
import requests
from dotenv import load_dotenv

```

```python
from langchain_groq import ChatGroq
from langchain.tools import tool
from langchain_community.tools.tavily_search import TavilySearchResults
from langsmith import Client  
```

```python
# agent related function not required create_react_agent and agent_executor for new langchain version

from langchain.agents import create_agent 
```

```python
os.environ["SSL_CERT_FILE"] = certifi.where()
load_dotenv()
```

```python
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
TAVILY_API_KEY = os.getenv("TAVILY_API_KEY")
```

```python
# first tool 
search_tool = TavilySearchResults(max_results =3 )
```

```python
response = search_tool.invoke("data science jobs for a fresher" )
```

```python
response 
```

```python
# the brain 

llm = ChatGroq(model = 'llama-3.3-70b-versatile', temperature =0, api_key = GROQ_API_KEY)
```

```python
tools = [search_tool]
```

```python
#agent creation

agent = create_agent(llm,tools,system_prompt= "you are a agent who uses the tools to answer the user queries and provide the best possible answer to the user in maximum 3 lines.")
```

```python
def agent_executor(query:str):
    response = agent.invoke({
        "messages" : [('user',query)]
    })
    return response['content']
```

```python
agent_executor("data science jobs for a fresher in india")
```
